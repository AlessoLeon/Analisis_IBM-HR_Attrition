```python
# Tamaño del grafico
plt.figure(figsize=(6,8))

# Titulo del Grafico
plt.title(
    "Rotación de Empleados",
    fontsize=16,
    fontweight="bold"
)

# Titulos para los Ejes
plt.xlabel("")
plt.ylabel("Cantidad de empleados")

# Tipo de Grafico, de donde lo extrae y que colocar en su Eje mas su color
ax = sns.countplot(
    data=df_spanish,
    x="Rotacion",
    hue="Rotacion",
    palette=["#e74c3c", "#2ecc71"],
    legend=False
)

# Total de Retación
total = len(df_spanish)

# Texto para las Barras (Yes | No)
for p in ax.patches:  #ax.patches contiene todas las barras dibujadas en el gráfico. (Recorrera Yes | No)
    conteo = p.get_height()  #Obtiene la altura de la barra actual. (Primero la altura de Yes | Luego la altura de No)
    porcentaje = 100 * conteo / total  #Calcula qué porcentaje representa esa barra respecto al total.

    # Anotaciones para las barras de los graficos
    ax.annotate(
        f"{int(conteo)}\n({porcentaje:.1f}%)", #Texto a mostrar en cada columna
        (p.get_x() + p.get_width()/2, conteo), # Coordenadas para colocar el texto (x,y) == (centro de la barra y en su altura máxima).
        ha='center', #Centra el texto encima de la columna
        va='bottom', #Coloca el texto encima de la columna
        fontsize=10, #Tamaño del texto encima de la columna
        fontweight='bold' #Negrita para el texto encima de la columna
    )

# Guardamos el grafico como PNG en la carpeta images del proyecto
plt.savefig(
    "../images/rotacion_empleados.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

# Muestra el Grafico "Rotación de Empleados"
plt.show()
```

```python
tasa_de_desgaste = (
    pd.crosstab(df_spanish['Horas_Extra'], df_spanish['Rotacion'], normalize='index') * 100
)

rate_yes = tasa_de_desgaste['Yes']

# --> Parte 1: La Hoja del Grafico (matplotlib.pyplot)
# Tamaño de la hoja grafica
plt.figure(figsize=(8,6))

plt.title(
    'Tasa de Rotación según Horas Extra',
    fontsize=18,
    weight='bold'
)

plt.ylabel('(%) de Empleados que Abandonaron')

plt.grid(axis='y', alpha=0.7)  #Opacidad para las lineas del plano grafico eje Y

# --> Parte 2: Diseño del grafico (matplotlib.pyplot)
# Histogramas/Graficas
# Grafico del tipo Bar (Grafico de Barra)
bars = plt.bar(
    rate_yes.index,  #Index de Si_Desgastados
    rate_yes.values,  #Valor de Si_Desgastados
    color=['#2A9D8F', '#E76F51']  #Color de Barras para plt.bar
)

# Ciclo para mostrar (%) de ambas barras
for bar in bars:
    y = bar.get_height()
    plt.text(  #Con esto escribimos lo que contine las barras del grafico
        bar.get_x() + bar.get_width()/2,
        y + 0.5,
        f'{y:.1f}%',
        ha='center',
        fontweight='bold'
    )

# Toque final y mostramos el grafico
plt.tight_layout()
plt.show()
```

### Hallazgos Principales

Los empleados que realizan horas extra tienen una probabilidad *significativamente mayor* de **abandonar** la empresa.

Según el gráfico:

- Sin horas extra: 10.4% de rotación.
- Con horas extra: 30.5% de rotación.

Esto significa que:

Los empleados que hacen horas extra tienen aproximadamente 3 veces más probabilidades de abandonar la empresa que aquellos que no las realizan.

### Observaciones:

1. La mayoría de empleados no hacen horas extra
1054 empleados no realizan horas extra.
416 empleados sí realizan horas extra.

Sin embargo...

2. Casi la misma cantidad de abandonos ocurre en ambos grupos
Sin horas extra: 110 abandonos.
Con horas extra: 127 abandonos.

Y eso es muy llamativo porque el grupo con horas extra es mucho más pequeño.

### Las horas extra parecen ser un factor de riesgo

Aunque solo 416 personas hacen horas extra:

- 127 abandonaron la empresa.
- Esto representa casi un tercio del grupo.

Mientras que en los empleados sin horas extra:

- Solo 110 de 1054 abandonaron.

La diferencia es enorme.

##### 📌 Descripción del Proyecto

La rotación de empleados representa uno de los mayores desafíos para las organizaciones, ya que impacta directamente en los costes, la productividad y la retención del talento.

En este proyecto se realizó un análisis completo del conjunto de datos "IBM HR Analytics Employee Attrition & Performance", con el objetivo de identificar los factores que influyen en la rotación de empleados y proporcionar información útil para la toma de decisiones en Recursos Humanos.

El análisis fue desarrollado utilizando Python para la limpieza, transformación y exploración de los datos, mientras que Power BI se utilizó para la construcción de un dashboard interactivo orientado al análisis ejecutivo.

##### 🎯 Objetivos del Proyecto
- Analizar los factores relacionados con la rotación de empleados.
- Identificar perfiles con mayor riesgo de abandono.
- Detectar patrones asociados a la satisfacción laboral.
- Evaluar el impacto del salario, edad, horas extras y experiencia laboral.
- Construir visualizaciones que faciliten la toma de decisiones basadas en datos.

- Explorar indicadores de desempeño y permanencia.

#### 📈 Principales Hallazgos

Entre los hallazgos más relevantes del análisis se identificó que:

✅ Los empleados que realizan horas extras presentan mayores tasas de rotación.

✅ Los trabajadores con menor antigüedad tienden a abandonar la empresa con más frecuencia.

✅ Determinados grupos de edad muestran una mayor propensión a la rotación.

✅ La satisfacción laboral y el equilibrio entre vida laboral y personal están relacionados con la permanencia del empleado.

✅ El nivel salarial puede influir en las decisiones de permanencia dentro de la organización.
