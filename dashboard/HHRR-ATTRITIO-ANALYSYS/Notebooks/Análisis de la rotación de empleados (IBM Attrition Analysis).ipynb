```python
sns.countplot(data=df_spanish,x='Rotacion')
plt.show()
```

```python
# Tamaño del grafico
plt.figure(figsize=(6,8))

# Titulo del Grafico
plt.title(
    "Rotación de Empleados",
    fontsize=16,
    fontweight="bold"
)

# Titulos para los Ejes
plt.xlabel("")
plt.ylabel("Cantidad de empleados")

# Tipo de Grafico, de donde lo extrae y que colocar en su Eje mas su color
ax = sns.countplot(
    data=df_spanish,
    x="Rotacion",
    hue="Rotacion",
    palette=["#e74c3c", "#2ecc71"],
    legend=False
)

# Total de Retación
total = len(df_spanish)

# Texto para las Barras (Yes | No)
for p in ax.patches:  #ax.patches contiene todas las barras dibujadas en el gráfico. (Recorrera Yes | No)
    conteo = p.get_height()  #Obtiene la altura de la barra actual. (Primero la altura de Yes | Luego la altura de No)
    porcentaje = 100 * conteo / total  #Calcula qué porcentaje representa esa barra respecto al total.

    # Anotaciones para las barras de los graficos
    ax.annotate(
        f"{int(conteo)}\n({porcentaje:.1f}%)", #Texto a mostrar en cada columna
        (p.get_x() + p.get_width()/2, conteo), # Coordenadas para colocar el texto (x,y) == (centro de la barra y en su altura máxima).
        ha='center', #Centra el texto encima de la columna
        va='bottom', #Coloca el texto encima de la columna
        fontsize=10, #Tamaño del texto encima de la columna
        fontweight='bold' #Negrita para el texto encima de la columna
    )

# Muestra el Grafico "Rotación de Empleados"
plt.show()
```

```python
# --> Parte 1: La Hoja del Grafico (matplotlib.pyplot)
# Tamaño de la hoja grafica
plt.figure(figsize=(12,7))

plt.title(
    'Horas Extra y Rotación de Empleados',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Realiza Horas Extra')
plt.ylabel('Cantidad de Empleados')

plt.xticks(
    ticks=[0, 1],
    labels=['Sí hizo Hras Extras', 'No hizo Hras Extras']
)

# --> Parte 2: Diseño del grafico (seaborn)
# Histogramas/Graficas
# Grafico del tipo Countplot (Grafico de Recuento)
ax = sns.countplot(
    data=df_spanish,
    x='Horas_Extra',
    hue='Rotacion',
    palette=['#2A9D8F','#E76F51']
)

# Texto mostrado en la leyenda
legend = ax.get_legend()
legend.set_title("Estado del Empleado:")
legend.texts[0].set_text("Si Abandono")
legend.texts[1].set_text("No Abandono")

sns.despine()  #Elimina algunos de los bordes (spines) del gráfico para que tenga un aspecto más limpio.
sns.set_style("darkgrid", {"grid.color": ".77"})  #Añade el estilo visual y la cuadrícula

# Etiquetas sobre cada barra
for container in ax.containers:
    ax.bar_label(container)

# Toque final y mostramos el grafico
plt.tight_layout()  #Ajusta automáticamente los espacios para que nada quede cortado
plt.show()  #Muestra el gráfico en pantalla
```

```python
tasa_de_desgaste = (
    pd.crosstab(df_spanish['Horas_Extra'], df_spanish['Rotacion'], normalize='index') * 100
)

rate_yes = tasa_de_desgaste['Yes']

# --> Parte 1: La Hoja del Grafico (matplotlib.pyplot)
# Tamaño de la hoja grafica
plt.figure(figsize=(8,6))

plt.title(
    'Tasa de Rotación según Horas Extra',
    fontsize=18,
    weight='bold'
)

plt.ylabel('(%) de Empleados que Abandonaron')

plt.grid(axis='y', alpha=0.7)  #Opacidad para las lineas del plano grafico eje Y

# --> Parte 2: Diseño del grafico (matplotlib.pyplot)
# Histogramas/Graficas
# Grafico del tipo Bar (Grafico de Barra)
bars = plt.bar(
    rate_yes.index,  #Index de Si_Desgastados
    rate_yes.values,  #Valor de Si_Desgastados
    color=['#2A9D8F', '#E76F51']  #Color de Barras para plt.bar
)

# Ciclo para mostrar (%) de ambas barras
for bar in bars:
    y = bar.get_height()
    plt.text(  #Con esto escribimos lo que contine las barras del grafico
        bar.get_x() + bar.get_width()/2,
        y + 0.5,
        f'{y:.1f}%',
        ha='center',
        fontweight='bold'
    )

# Toque final y mostramos el grafico
plt.tight_layout()
plt.show()
```

### Hallazgos Principales

Los empleados que realizan horas extra tienen una probabilidad *significativamente mayor* de **abandonar** la empresa.

Según el gráfico:

- Sin horas extra: 10.4% de rotación.
- Con horas extra: 30.5% de rotación.

Esto significa que:

Los empleados que hacen horas extra tienen aproximadamente 3 veces más probabilidades de abandonar la empresa que aquellos que no las realizan.

### Observaciones:

1. La mayoría de empleados no hacen horas extra
1054 empleados no realizan horas extra.
416 empleados sí realizan horas extra.

Sin embargo...

2. Casi la misma cantidad de abandonos ocurre en ambos grupos
Sin horas extra: 110 abandonos.
Con horas extra: 127 abandonos.

Y eso es muy llamativo porque el grupo con horas extra es mucho más pequeño.

### Las horas extra parecen ser un factor de riesgo

Aunque solo 416 personas hacen horas extra:

- 127 abandonaron la empresa.
- Esto representa casi un tercio del grupo.

Mientras que en los empleados sin horas extra:

- Solo 110 de 1054 abandonaron.

La diferencia es enorme.

### 💼 Perspectiva de negocio

Desde Recursos Humanos, este resultado podría indicar:

✅ Sobrecarga laboral.\
✅ Estrés acumulado.\
✅ Menor equilibrio vida-trabajo.\
✅ Mayor agotamiento (burnout).\
✅ Menor satisfacción laboral.

Todo ello podría estar contribuyendo al aumento de la rotación.

```python
# Calcular porcentaje de rotación por departamento
rotacion_dep = (
    df.groupby('Department')['Attrition']
      .value_counts(normalize=True)
      .mul(100)
      .rename('Porcentaje')
      .reset_index()
)

# Solo empleados que abandonaron
rotacion_dep = rotacion_dep[rotacion_dep['Attrition'] == 'Yes']

# Ordenar de mayor a menor
rotacion_dep = rotacion_dep.sort_values(
    by='Porcentaje',
    ascending=True
)

# --> Parte 1: La Hoja del Grafico (matplotlib.pyplot)
# Tamaño del gráfico
plt.figure(figsize=(12,6))

# Título
plt.title(
    'Probabilidad de Rotación según Departamento',
    fontsize=18,
    fontweight='bold',
    pad=15
)

# Ejes
plt.xlabel('(%) de Empleados que Abandonaron', fontsize=13)
plt.ylabel('Departamento', fontsize=13)

# --> Parte 2: Diseño del grafico (countplot)
# Gráfico
ax = sns.barplot(
    data=rotacion_dep,
    x='Porcentaje',
    y='Department',
    hue='Department',
    palette='viridis',
    legend=False
)

# Traducir departamentos
ax.set_yticks(ax.get_yticks())
ax.set_yticklabels([
    'Ventas',
    'Recursos\nHumanos',
    'Investigación\ny Desarrollo'
])

# Etiquetas en barras
for i, valor in enumerate(rotacion_dep['Porcentaje']):
    plt.text(
        valor + 0.3,
        i,
        f'{valor:.2f}%',
        va='center',
        fontweight='bold'
    )

# Retoques y mostrar el grafico
sns.despine()
plt.tight_layout()
plt.show()
```

### Conclusion

Se aprecia que:

- **Investigación y Desarrollo (I+D)** concentra la mayor cantidad de empleados. (828 Empleados)
- **Ventas** es el segundo departamento con más personal. (354 Empleados)
- **Recursos Humanos (RRHH)** tiene una plantilla considerablemente menor. (51 Empleados)

Esto indica que la empresa está principalmente enfocada en actividades de investigación y desarrollo, siendo este el núcleo de la organización.

### Porcentaje de empleados que abandonaron la empresa

Departamentos:

- _Investigación y Desarrollo:_ De 828 Empleados, abandonaron 133, lo que indica que de ese departamento abandono un **(25,63%)** de empleados.
- _Ventas:_ De 354 Empleados, abandonaron 92, lo que indica que de ese departamento abandono un **(13,84%)** de empleados.
- _Recursos Humanos:_ De 51 Empleados, abandonaron 12, lo que indica que de ese departamento abandono un **(19,05%)** de empleados.

### Hallazgos

La rotación no afecta por igual a todos los departamentos. 

- Recursos Humanos presenta el mayor riesgo proporcional de abandono
- Mientras que Investigación y Desarrollo concentra el mayor número total de empleados que dejan la empresa.

Esto sugiere que las estrategias de retención deberían diseñarse de forma específica para cada departamento.
